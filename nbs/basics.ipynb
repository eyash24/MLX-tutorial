```python
# Accessing arrays 

a = mx.array([10,20,30,40,50])

for x in range(a.size):
    a[x] = a[x]+ 5
    print(a[x])

print(a)
```

```python
a = mx.array([10,20,30,40,50])

b = mx.add(a, 5)
c = mx.subtract(a, 3)
d = mx.multiply(a, 7)
e = mx.divide(a, 3)

print(a)
print(b)
print(c)
print(d)
print(e)
```

```python
a = mx.array([10.42,20.90,30.75,40.12,50.36])

b = mx.add(a, 5)
c = mx.subtract(a, 3)
d = mx.multiply(a, 7)
e = mx.divide(a, 3)

print(a)
print(b)
print(c)
print(d)
print(e)
```

```python
a = mx.array([10,20,30,40,50])

b = mx.sum(a)
c = mx.min(a)
d = mx.max(a)
e = mx.mean(a)
f = mx.std(a)

print(a)
print(b)
print(c)
print(d)
print(e)
print(f)
```

```python
a = mx.array([[10], [20], [30], [40], [50]])
b = mx.array([[60], [70], [80], [90], [100]])

c = mx.add(a,b)
d = mx.subtract(a,b)
e = mx.multiply(a,b)
f = mx.divide(a,b)

print(a)
print(b)
print(c)
print(d)
print(e)
print(f)
```

```python
a = mx.array([10,20,30,40,50])
b = mx.array([60, 70, 80, 90, 100])

c = mx.add(a,b)
d = mx.subtract(a,b)
e = mx.multiply(a,b)
f = mx.divide(a,b)

print(a)
print(b)
print(c)
print(d)
print(e)
print(f)
```

```python
# dot product
a = mx.array([1,2,3], mx.float32)
b = mx.array([4,5,6], mx.float32)

c = mx.matmul(a,b)

print(a)
print(b)
print(c)
```

```python
# dot product
a = mx.array([[1,2], [3, 4]], mx.float32)
b = mx.array([[4,5],[6,7]], mx.float32)

c = mx.matmul(a,b)

print(a)
print(b)
print(c)
```

```python
# Shaping 

a = mx.arange(10, 60, 10)

print(a)
```

```python
a = mx.zeros([5,1])

print(a)
```

```python
a = mx.arange(10,60,10)
b = mx.zeros(a.shape)

print(a)
print(b)
```

```python
c = mx.random.randint(0, 100, a.shape)
print(c)
```

```python
a = mx.array([[[10,20,30], [40, 50, 60]]])
b = mx.flatten(a)
c = mx.squeeze(a)
d = mx.expand_dims(c, 1)

print(a)
print(b)
print(c)
print(d)
```

```python
a = mx.array([1,20,30])
b = mx.array([40,50,60])

c = mx.concatenate([a,b])

print(a)
print(b)
print(c)
```

```python
a = mx.array([10,20,30,40,50])

b = mx.reshape(a, (5,1))

print(a)
print(b)
```

```python
a = mx.array([10,20,30])

b = mx.broadcast_to(a, (3,3))

print(a)
print(b)
```

```python
a = mx.array([[10,20,30], [40, 50, 60]])

b = mx.transpose(a)

print(a)
print(b)
```

```python
a = mx.array([[10,20,30], [40, 50, 60]])
b = mx.array([[10,20,30], [40, 50, 60]])

c = mx.add(a,b, stream=mx.cpu)
d = mx.add(a,b, stream=mx.gpu)

print(a)
print(b)
print(c)
print(d)
```
